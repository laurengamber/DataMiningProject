```python
#Load Packages
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
```

```python
# Read in
fairy_tales = pd.read_csv("grimms_fairytales.csv")
fairy_tales.head(10)
```

```python
# Clean the text and create a document-term matrix of word counts
vectorizer = CountVectorizer(
    lowercase=True,
    stop_words="english",
    token_pattern=r"\b[a-zA-Z]{3,}\b"  # letters only, 3+ characters
)
```

```python
# Clean the text and create a document-term matrix of word counts
vectorizer = CountVectorizer(
    lowercase=True,
    stop_words="english",
    token_pattern=r"\b[a-zA-Z]{3,}\b"  # letters only, 3+ characters
)

dtm = vectorizer.fit_transform(fairy_tales["Text"].astype(str))
```

```python
#Fit LDA Model
lda_fit = LatentDirichletAllocation(n_components=5, random_state=123)
lda_fit.fit(dtm)
```

```python
# Show the 10 highest-probability words in each topic
words = vectorizer.get_feature_names_out()

topic_word = pd.DataFrame(lda_fit.components_, columns=words,
                          index=["Topic 1", "Topic 2", "Topic 3", "Topic 4", "Topic 5"])

terms = topic_word.apply(lambda row: row.nlargest(10).index.tolist(),
                         axis=1, result_type="expand").T
print(terms)
```
