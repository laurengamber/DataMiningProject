```python
protected = json.loads((OUT/'protected_inputs.json').read_text())
archive = OUT/'label_confirmation_archive/first_three/original_locations.json'
archived = json.loads(archive.read_text()) if archive.exists() else {}
allowed_status = {'comparison_evidence_checklist.csv', 'completion_plan.csv'}
# Approved label updates have original copies; rebuilt comparison has an archived original.
changed = []
for path, expected in protected.items():
    preserved_path = REPO/archived.get(path, path)
    if Path(path).name not in allowed_status and checksum(preserved_path) != expected:
        changed.append(path)
assert not changed, changed
reservation = json.loads((MODERN/'Results/modern/review_sample.json').read_text())
review = pd.read_csv(OUT/'label_review/topic_review_index.csv')
evidence_ids = {x.strip() for s in review.evidence_ids for x in s.split('|')}
assert not evidence_ids.intersection(reservation['article_ids'])
decisions = pd.read_csv(OUT/'label_review/team_decisions.csv', keep_default_na=False)
print('Original artifacts preserved, approved label inputs archived before edits, no reserved development evidence.')
print('Explicit assisted label approvals:', decisions.decision.eq('approve').sum(), 'of 110. Independent ratings remain pending.')
```
